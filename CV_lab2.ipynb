```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
def show_image(title,img,cmap=None):
    plt.figure(figsize=(10,10))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    if cmap:
      plt.imshow(img,cmap=cmap)
    else:
      plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
```

```python
img1 = cv2.imread("/content/drive/MyDrive/marvels-spider-man-3840x2160-11990.jpeg")
show_image("a",img1)
```

```python
img2=cv2.imread("/content/drive/MyDrive/Michael_Jackson_-_Thriller.png")
show_image("b",img2)
```

```python
print(img1.shape)
print(img2.shape)
```

```python
img1=cv2.resize(img1,(img2.shape[1],img2.shape[0]))
print(img2.shape)
```

```python
show_image("a",img1)
```

```python
#addition
add=cv2.addWeighted(img1,0.5,img2,0.5,0)
show_image("arithmetic addition(Blending)",add)
```

```python
#subtaction
sub=cv2.subtract(img1,img2)
show_image("arithmetic subtaction",sub)
```

```python
#create a rectangle to demo bitwise ops
rectangle=(np.zeros((img1.shape[0],img1.shape[1]),dtype="uint8"))
cv2.rectangle(rectangle,(50,50),(300,300),255,-1)
show_image("rectangle",rectangle)
```

```python
#create circle
circle=np.zeros((img1.shape[0],img1.shape[1]),dtype="uint8")
cv2.circle(circle,(200,200),200,255,-1)
show_image("circle",circle)
```

```python
#bitwise OR
bitwise_or=cv2.bitwise_or(rectangle,circle)
show_image("bitwise_or",bitwise_or,cmap="gray")
```

```python
#bitwise XOR
bitwise_xor=cv2.bitwise_xor(rectangle,circle)
show_image("bitwise_xor",bitwise_xor,cmap="gray")
```

```python
#bitwise NOT
bitwise_not=cv2.bitwise_not(circle)
show_image("bitwise_not",bitwise_not,cmap="gray")
```

```python
#image rotation clockwise 90
height,width,_=img1.shape
rotate_clockwise=np.zeros((width,height,3),dtype="uint8")
for i in range(height):
  for j in range(width):
    rotate_clockwise[j,height-i-1]=img1[i,j]
show_image("rotation by 90(clockwise)",rotate_clockwise)
```

```python
#image rotation 90 anticlockwise
height,width,_=img1.shape
rotate_anticlockwise=np.zeros((width,height,3),dtype="uint8")
for i in range(width):
  for j in range(height):
    rotate_anticlockwise[i,j]=img1[j,width-i-1]
show_image("rotation by 90(anticlockwise)",rotate_anticlockwise)
```
